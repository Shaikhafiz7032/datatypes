```python
#Question 49: Convert 'hello world' to ROT13 encoding")
text = "hello world"
alphabet = "abcdefghijklmnopqrstuvwxyz"
rot = "nopqrstuvwxyzabcdefghijklm"
result = ""
for ch in text:
    if ch in alphabet:
        index = alphabet.index(ch)
        result += rot[index]
    else:
        result += ch

print(result)
```

```python
#Question 50: Check if string is a valid credit card number: '4532015112830366'
num = "4532015112830366"
if num.isdigit() and len(num) >= 13 and len(num) <= 19:
    total = 0
    num = num[::-1]   # reverse
    for i in range(len(num)):
        digit = int(num[i])
        if i % 2 == 1:
            digit = digit * 2
            if digit > 9:
                digit = digit - 9
        total = total + digit

    if total % 10 == 0:
        print("Valid card")
    else:
        print("Invalid card")

else:
    print("Invalid card")
```
