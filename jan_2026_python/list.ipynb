```python
#Question 10: Create a list of prime numbers between 1 and 50")
lis = []
for num in range(1,50):
    is_prime=True
    for i in range(2,num):
        if num% i == 0 :
            is_prime=False
            break
    if is_prime:
        lis.append(num)
print(lis)

    
```

```python
#Question 11: Flatten nested list [[1, 2, 3], [4, 5, 6], [7, 8, 9]]")
lis =  [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
new=[]
for num in lis:
    for i in num:
        if i not in new:
            new.append(i)
print(new)
```

```python
#Question 18: Find the sum of all elements in 3D list [[[1, 2], [3, 4]], [[5, 6], [7, 8]]]")
lis1 = [[[1, 2], [3, 4]], [[5, 6], [7, 8]]]
total =0 
for layers in lis1:
    for row in layers:
        for num in row:
            total = total+ num
print(total)

```
